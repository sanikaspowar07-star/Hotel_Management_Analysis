```python
#import libraries 
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#load Dataset 
df=pd.read_csv(r"C:\Users\Lenovo\Downloads\0_Hotel_management Data.csv")
df.head()
```

```python
#remove the unneccsssary Column 
df=df.drop(columns=["Column1"])
df.head()
```

```python
#Check Dataset Shape 
print("Numbers of Rows:",df.shape[0])
print("Numbers of Rows:",df.shape[1])
```

```python
#Display Column name 
print(df.columns.tolist())
```

```python
#Check Missing Value 
print(df.isnull().sum())
```

```python
#Check Duplicate Records 
print("Total duplicate Rows:")
df.duplicated().sum()
```

```python
#Check Data Types 
df.dtypes
```

```python
#Statstical Summary 
df.describe()
```

```python
#Explore Categorical Columns 
print(df.columns.tolist())
```

```python
print(df["Booking Status"].value_counts())
```

```python
#Analyze customer Type 
print(df["Customer Type"].value_counts())
```

```python
#Analyze booking Status 
booking_percentage=df["Booking Status"].value_counts(normalize=True)*100
print(booking_percentage.round(2))
```

```python
#Analyze Total revenue by Room Type 
revenue_by_room=df.groupby("Room Type")["Total Revenue"].sum()
print(revenue_by_room)
```

```python
#Analyze Booking Status room type 
booking_by_room=pd.crosstab(df["Room Type"],df["Booking Status"])
print(booking_by_room)
```

```python
#most Popular Room Type
room_counts=df["Room Type"].value_counts()
print(room_counts)
```

```python
#most Popular room type 
print("Most Popular Room Type:",room_counts.idxmax())
print("Total Bookings:",room_counts.max())
```

```python
#Analyze Revenue By City 
revenued_by_city=df.groupby("City")["Total Revenue"].sum()
print(revenued_by_city)
```

```python
#Analyze Revenue by Payment Mode 
revenue_by_payment=df.groupby("Payment Mode")["Total Revenue"].sum()
print(revenue_by_payment)
```

```python
#Analyze bookings by City
bookings_by_city=df["City"].value_counts()
print(bookings_by_city)
```

```python
#City With the most Booking
print("City with the most bookings:",bookings_by_city.idxmax())
print("Total bookings:",bookings_by_city.max())
```

```python
#Analyze Discounts 
average_discount=df["Discount"].mean()
print("Average Discount:",round(average_discount,2))
```

```python
#Highest Discount 
print("Highest Discount:",df["Discount"].max())
print("Lowest Discount:",df["Discount"].min())
```

```python
#Analyze Room Availability 
print(df["Room Available"].value_counts())
```

```python
#Analyze Customer age 
print("Average Customer Age:",round(df["Customer Age"].mean(),2))
print("Minimum Customer Age:",df["Customer Age"].min())
print("Maximum Customer Age:",df["Customer Age"].max())
```

```python
#Analyze Customer Gender 
print(df["Customer Gender"].value_counts())
```

```python
#Visualize Customer Gender 
df["Customer Gender"].value_counts().plot(kind="bar",figsize=(5,4))
plt.title("Customer Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Customers")
plt.show()
```

```python
#analyze Booking Status By Customer type 
booking_analysis=pd.crosstab(df["Customer Type"],df["Booking Status"])
booking_analysis.plot(kind="bar",figsize=(6,5))
plt.title("Booking Status by Customer Type")
plt.xlabel("Customer Type")
plt.ylabel("Number of Bookings")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
#Analyze Booking status Using Piechart
df["Booking Status"].value_counts().plot(kind="pie",autopct="%1.1f%%",figsize=(4,4))
plt.title("Booking Status Distribution")
plt.ylabel("")
plt.show()
```

```python
plt.figure(figsize=(5, 3))
plt.hist(df["Customer Age"], bins=10, edgecolor="black")
plt.title("Customer Age Distribution", fontsize=10)
plt.xlabel("Customer Age", fontsize=9)
plt.ylabel("Number of Customers", fontsize=9)
plt.xticks(fontsize=8)
plt.yticks(fontsize=8)
plt.tight_layout()
plt.show()
```

```python
#Analyze Revenue by Customer Type
revenue_by_customer = df.groupby("Customer Type")["Total Revenue"].sum()
print(revenue_by_customer)
```

```python
#Visualize Revenue by Customer Type 
revenue_by_customer.plot(kind="bar", figsize=(5, 4), edgecolor="black")
plt.title("Revenue by Customer Type", fontsize=10)
plt.xlabel("Customer Type", fontsize=9)
plt.ylabel("Total Revenue", fontsize=9)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
#Top 5 Cities BY Bookings
city_bookings = df["City"].value_counts().head(5)
city_bookings.plot( kind="bar",figsize=(4, 3), edgecolor="black")
plt.title("Top 5 Cities by Bookings", fontsize=10)
plt.xlabel("City", fontsize=9)
plt.ylabel("Number of Bookings", fontsize=9)
plt.xticks(rotation=45, fontsize=8)
plt.tight_layout()
plt.show()
```

```python
#Payment mode Visualization
df["Payment Mode"].value_counts().plot(kind="pie",autopct="%1.1f%%",figsize=(4, 4))
plt.title("Payment Mode Distribution")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

```python
#Discount Visualization 
plt.figure(figsize=(4, 3))
plt.hist(df["Discount"], bins=10, edgecolor="black", rwidth=0.85)
plt.title("Discount Distribution", fontsize=10)
plt.xlabel("Discount", fontsize=9)
plt.ylabel("Number of Bookings", fontsize=9)
plt.tight_layout()
plt.show()
```

```python
#Customer Age Vs Total Revenue
sample_df = df.sample(n=1000, random_state=42)
plt.figure(figsize=(6, 4))
plt.scatter( sample_df["Customer Age"],sample_df["Total Revenue"],alpha=0.5, s=15)
plt.title("Customer Age vs Total Revenue", fontsize=11)
plt.xlabel("Customer Age", fontsize=9)
plt.ylabel("Total Revenue", fontsize=9)
plt.grid(True, alpha=0.2)
plt.tight_layout()
plt.show()
```
